## Теорема Фишера для музыкальных жанров

В книге «Эгоистичный ген» Ричард Докинз высказал гипотезу о том, что эволюция идей в человеческой культуре происходит идентично эволюции биологической, и предложил понятие _мема_ как культурного аналога гена -- наименьшей биологической единицы, способной к репликации. Идея Докинза является, в сущности, лишь метафорой, однако она открывает возможность для исследования культурных явлений методами биологии. В данной работе мы проводим исследование в таком духе, а именно, пытаемся выяснить, выполняется ли теорема Фишера для «популяций» музыкальных жанров. 

Теорема Фишера о естественном отборе -- один из фундаментальных результатов математической популяционной биологии, сформулированный Рональдом Фишером в 1930 году в книге «Генетическая теория естественного отбора». Формулируется эта теорема так: 

>_Скорость повышения приспособленности в любой популяции в любой момент равна её генетической дисперсии по приспособленности в этот момент_ (Солбриг О., Солбриг Д. Популяционная биология и эволюция. М, Мир, 1982. - стр. 163).

Существует несколько разных версий конкретных математических выражений теоремы Фишера. В частности, можно сформулировать её как для диплоидных организмов, которые размножаются половым путем, так и для гаплоидных, размножающихся делением. Если смотреть на музыкальный жанр как на популяцию особей-треков, будет довольно трудно придать смысл идее «скрещивания треков», поэтому ниже мы формулируем и доказываем теорему Фишера в её гаплоидном варианте. О том, как можно определить «геном трека», мы поговорим ниже. 

## Формулировка и доказательство теоремы Фишера 

Исходные допущения таковы. Мы рассматриваем два момента времени, $t_0$ и $t_1$. Мы считаем, что в момент $t_0$ существовало одно поколение, каждая особь которого обладала одним из $N$ различных генотипов. Оно размножилось, и в момент $t_1$ мы наблюдаем следующее поколение, не пересекающееся с первым, и имеющее другое распределение тех же генотипов. Обозначим через $p_i$ долю $i$-того генотипа в поколении $t_0$, через $q_i$ -- долю $i$-того генотипа в поколении $t_1$ (в частности, $\sum_j^N p_j = \sum_j^N q_j = 1$). 

Для каждого генотипа $i$ введем его относительную приспособленность $w_i$ и определим среднюю 
приспособленность формулой 
$$\bar W_0 = \sum_j p_jw_j$$
Наша задача -- выразить изменение этой величины в терминах приспособленностей и начального распределения. После отбора, в новом поколении число особей с данным генотипом пропорционально приспособленности и исходной вероятности, то есть
$$q_i = \frac{p_iw_i}{\bar W_0}.$$
В предположении, что приспособленность в течение одного поколения не поменялась, средняя приспособленность в момент $t_1$ рассчитывается по формуле $\bar W_1 = \sum_jq_jw_j$. Тогда имеем
$$\Delta \bar W = \bar W_1 - \bar W_0 = \sum_j(q_j - p_j)w_j = 
\sum_j\Big(\frac{p_jw_j}{\bar W_0} - p_j\Big)w_j = \frac{1}{\bar W_0} \sum_j p_j(w_j^2 - w_j\bar W_0)=
$$
$$ = \frac{1}{\bar W_0} \sum_j p_j(w_j - \bar W_0)^2.$$
В последнем шаге мы воспользовались тождеством 
$$\sum_j p_j(-w_j\bar W_0 + \bar W_0^2) = \bar W_0 \sum_j (-p_jw_j + p_j\bar W_0) = 
\bar W_0  (-\bar W_0 + \sum_j p_j\bar W_0) = \bar W_0 (-\bar W_0 + \bar W_0) = 0.$$
Величина $\sum_j p_j(w_j - \bar W_0)^2$ называется генетической дисперсией по приспособленности, тем самым теорема доказана. 




## Как применить теорему Фишера к музыкальному жанру?   

Для эмпирической проверки теоремы Фишера необходимо ввести определения генотипов и приспособленностей для конкретных популяций. Изложим их, следуя цитированной выше книге Солбригов.
Как и раньше, мы предполагаем, что есть $N$ генотипов, и два поколения $t_0,t_1$ численностью $P,Q$. 

- Пусть $P_i$ -- численность $i$-го генотипа (так что $\sum_j P_j = P$). 

- Аналогично $Q_j$ -- численность $i$-го генотипа в поколении $Q$.

- Имеем $p_i = P_i / P, q_i = Q_i / Q$. 

- Определим выживаемость $\lambda_i = Q_i / P_i$;

- $\Lambda = \max_j \lambda_j$;

-  Относительная приспособленность $w_i = \lambda_i / \Lambda = Q_i / P_i\Lambda$. Нормировка на максимальную выживаемость позволяет исключить влияние факторов, одноврменно влияющих на все генотипы, и рассматривать непосредственно отбор по генам. 

- Средняя приспособленность $\bar W_0 = \sum_j p_j w_j = \sum_j \frac{P_j}{P}\cdot \frac{Q_j}{P_j\Lambda} = \frac{Q}{P\Lambda}$. 

Предпосылкой теоремы Фишера является равенство $q_i = \frac{p_iw_i}{\bar W_0}$. Однако при таком определении величин оно является тождеством, поскольку
$$\frac{p_iw_i}{\bar W_0} = \frac{P_i}{P}\cdot\frac{Q_i}{P_i\Lambda}\cdot\frac{P\Lambda}{Q} = \frac{Q_i}{Q} = q_i.$$
Поэтому, чтобы проверка теоремы была нетриваильной, необходимо вычислять разность приспособленностей относительно трех поколений. Именно, введем третий момент времени $t_2$ и третье распределение численностей генотипов $S_i, \sum_j S_j = S$. Положим

- $\bar W_0 = \frac{Q}{P\Lambda_{P,Q}}, \bar W_1 = \frac{S}{Q\Lambda_{Q,S}}$

- Тестируемая гипотеза: для поколений $t_0,t_2, t_2, t_3...$ величины $\Delta \bar W(t_i) = \bar W(t_{i+1}) - \bar W(t_i)$ и $\frac{1}{\bar W(t_i)} \sum_j p_j(w_j(t_i) - \bar W(t_i))^2$ коррелируют.  

Нам осталось определить, что такое генотип музыки. Мы используем для этого датасет `rohiteng/spotify-music-analytics-dataset`. Помимо технических track_id, album_name, track_name, artist_name, duration_ms, country, stream_count, label, он содержит следующие колонки:

- track_genre: жанр трека (позволит разделить треки на «популяции»)

- release_date: дата выпуска (позволит разделить треки на «поколения»)

- explicit: эксплицитная лирика (True если да, False если нет)

- danceability: танцевабельность (от 0.0 до 1.0)

- energy: энергичность, от 0.0 до 1.0

- key: ключ (от 0 до 11)

- loudness: громкость в децибелах (dB)

- mode: модальность (1, если мажор, 0, если минор)

- instrumentalness: инструментальность трека (от 0.0 до 0.8)

- tempo: BPM 

- popularity (от 0 до 100)

Используя эти данные, мы должны составить генотип для каждого трека. Мы можем отбросить популярность, так как она должна быть зависимой от генотипа, и громкость, так как её регуоирует слушатель. Мы также отбросим темп, ибо его трудно бинаризовать, и ключ. Из оставшихся параметров explicit и mode уже являются бинарными признаками, a dancebility, energy и instrumentalness мы бинаризуем, беря порог >= 0.5 (>= 0.4 для инструментальности). Итого у нас получается 5 «генов» с двумя аллелями, всего $2^5 = 32$ варианта генотипов. 


## Разведочный анализ

```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv('./data/spotify-2025/spotify_2015_2025_85k.csv')

df["release_date"] = pd.to_datetime(df["release_date"])
df['year'] = df["release_date"].dt.year

print(f'Общее количество треков: {df.shape[0]}')
print(f'Количество жанров: {df['genre'].nunique()}')
print(f'Количество лет: {df['year'].nunique()}')
```

```python
# Распределение треков по годам и по жанрам

tracks_per_year = df["year"].value_counts().sort_index()
tracks_per_genre = df["genre"].value_counts().sort_values(ascending=False)

fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(14, 6))

# Распределение по годам (ax1)
tracks_per_year.plot(kind="bar", color="skyblue", edgecolor="black", ax=ax1)
ax1.set_title("Количество треков по годам")
ax1.set_xlabel("Год")
ax1.set_ylabel("Количество треков")
ax1.set_xticklabels(ax1.get_xticklabels(), rotation=45)
ax1.grid(axis="y", linestyle="--", alpha=0.7)

# Распределение по жанрам (ax2)
tracks_per_genre.plot(kind="bar", color="lightgreen", edgecolor="black", ax=ax2)
ax2.set_title("Количество треков по жанрам")
ax2.set_xlabel("Жанр")
ax2.set_ylabel("Количество треков")
ax2.set_xticklabels(ax2.get_xticklabels(), rotation=45)
ax2.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
# Приведем нужные нам колонки к правильному виду и создадим колонку с генотипом.

df['mode'] = df['mode']
df['explicit'] = (df['explicit']).astype(int)
df['energy'] = (df['energy'] >= 0.5).astype(int)
df['danceability'] = (df['danceability'] >= 0.5).astype(int)
df['instrumentalness'] = (df['instrumentalness'] >= 0.4).astype(int)

df['genotype'] = (
    df['mode'] * 16 +
    df['explicit'] * 8 +  
    df['energy'] * 4 +
    df['danceability'] * 2 +
    df['instrumentalness']
)

print(f'Число разных генотипов: {df['genotype'].nunique()}')
```

```python
# Распределение разных генотипов

tracks_per_genotype = df["genotype"].apply(lambda x: f"{x:b}").value_counts().sort_values(ascending=True)

tracks_per_genotype.plot(kind="bar", color="lightgreen", edgecolor="black")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.xlabel('Генотипы')
plt.ylabel('Количество треков')

plt.tight_layout()
plt.show()
```

Интересно, что генотипы делятся на две группы по 16 с примерно одинаковым количеством треков в каждой. Судя по графику, за это отвественен ген explicit. Чтобы проверить это формально, используем тест $\chi^2$ для проверки равномерности распределения каждого гена (и мы видим, что максимальное значение $\chi^2$ достигается на explicit). 

```python
from scipy.stats import chisquare

genes = ["mode", "explicit", "energy", "danceability", "instrumentalness"]

print("Доля аллелей '1' среди генов ")
for gene in genes:
    share = df[gene].mean()  # Так как значения 0 и 1, среднее равно доле единиц
    print(f"Ген {gene:16}: {share:.4f} (аллель 0: {1-share:.4f})")

print("\nТест хи-квадрат на равномерность (отклонение от 50/50)")
for gene in genes:
    counts = df[gene].value_counts()
    f_obs = counts.values
    f_exp = [len(df) / 2, len(df) / 2]
    stat, p_val = chisquare(f_obs, f_exp)
    print(f"Ген {gene:16}: Chi2-stat = {stat:10.1f}, p-value = {p_val:.4e}")
```

## Проверка гипотезы

Мы проверим гипотезу сначала на отдельных жанрах, потом на всем датасете, и в конце объединим данные по всем жанрам. Для оценки корреляции мы будем использовать классический коэффициент корреляции Пирсона, и дополнительно коэффициент Спирмена, поскольку зависимость может оказаться нелинейной. 


```python
# Определим основные функции

def survivability(year: int, df):
    base = df[df["year"] == year]["genotype"].value_counts()
    next_gen = df[df["year"] == year + 1]["genotype"].value_counts()

    lambdas_series = next_gen.divide(base, fill_value=0)
    full_index = range(32)
    return lambdas_series.reindex(full_index, fill_value=0.0).tolist()


def delta_fitness(year: int, df):
    lambdas = survivability(year, df)
    max_surv = max(lambdas)
    if max_surv == 0:
        return 0.0

    Q = df[df["year"] == year + 1].shape[0]
    P = df[df["year"] == year].shape[0]
    if P == 0:
        return 0.0

    return Q / (P * max_surv)


def fitness_variance_fast(year: int, df):
    df_year = df[df["year"] == year]
    P = len(df_year)
    if P == 0:
        return 0.0

    W_0 = delta_fitness(year, df)
    if W_0 == 0:
        return 0.0

    lambdas = survivability(year, df)
    max_surv = max(lambdas)
    if max_surv == 0:
        return 0.0

    w_series = pd.Series([l / max_surv for l in lambdas], index=range(32))
    p_series = (
        df_year["genotype"].value_counts().reindex(range(32), fill_value=0)
    )
    p_series = p_series / P

    variance = (p_series * ((w_series - W_0) ** 2)).sum()
    return variance / W_0
```

```python
# Проверим на жанре Classical

from scipy.stats import spearmanr, pearsonr
import numpy as np

sub_df = df[df["genre"] == "Classical"]

deltas = []
variances = []

for year in range(2015, 2025):
    deltas.append(delta_fitness(year, sub_df))
    variances.append(fitness_variance_fast(year, sub_df))


# 2. Линейная корреляция Пирсона
r_pearson, p_pearson = pearsonr(deltas, variances)

# Проверка гипотезы тестом Спирмена
rho, p_val = spearmanr(deltas, variances)


print(f"Результаты теста для жанра Classical")
print(f"Ранговая корреляция Пирсона: {r_pearson:.4f}")
print(f"Статистическая значимость (p-value): {p_pearson:.4f}")
print(f"Ранговая корреляция Спирмена: {rho:.4f}")
print(f"Статистическая значимость (p-value): {p_val:.4f}")

# Визуализация

slope, intercept = np.polyfit(deltas, variances, 1) # Находим коэффициенты прямой y = kx + b
x_range = np.linspace(min(deltas), max(deltas), 100)
plt.plot(x_range, slope * x_range + intercept, color='blue', linestyle='-', linewidth=2)

plt.scatter(deltas, variances)
#plt.axline((0, 0), (1, 1), color="red", linestyle="--", label="Линия Фишера (1:1)")
plt.xlabel("Delta Fitness")
plt.ylabel("Fitness Variance")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

plt.show()
```

```python
# Проверим на жанре Metal

sub_df = df[df["genre"] == "Metal"]

deltas = []
variances = []

for year in range(2015, 2025):
    deltas.append(delta_fitness(year, sub_df))
    variances.append(fitness_variance_fast(year, sub_df))


# 2. Линейная корреляция Пирсона
r_pearson, p_pearson = pearsonr(deltas, variances)

# Проверка гипотезы тестом Спирмена
rho, p_val = spearmanr(deltas, variances)


print(f"Результаты теста для жанра Classical")
print(f"Ранговая корреляция Пирсона: {r_pearson:.4f}")
print(f"Статистическая значимость (p-value): {p_pearson:.4f}")
print(f"Ранговая корреляция Спирмена: {rho:.4f}")
print(f"Статистическая значимость (p-value): {p_val:.4f}")

# Визуализация

slope, intercept = np.polyfit(deltas, variances, 1) # Находим коэффициенты прямой y = kx + b
x_range = np.linspace(min(deltas), max(deltas), 100)
plt.plot(x_range, slope * x_range + intercept, color='blue', linestyle='-', linewidth=2)

plt.scatter(deltas, variances)
#plt.axline((0, 0), (1, 1), color="red", linestyle="--", label="Линия Фишера (1:1)")
plt.xlabel("Delta Fitness")
plt.ylabel("Fitness Variance")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

plt.show()
```

Десять точек недостаточно для проверки корреляций, поэтому в качестве периода поколения возьмем кварталы, но проверим гипотезу не на отдельном жанре, а на всем датасете.  

```python
# 2. Создаем сквозной индекс квартала (например, с 2015 года)
# Формула: (Год - 2015) * 4 + Номер_Квартала
df['quarter_idx'] = (df['release_date'].dt.year - 2015) * 4 + df['release_date'].dt.quarter

# Посмотрим на доступный диапазон кварталов
# Для периода 2015-2024 годов индексы будут примерно от 1 до 40
min_q = int(df['quarter_idx'].min())
max_q = int(df['quarter_idx'].max())

def survivability_quarterly(q_idx: int, df):
    # Фильтруем данные по текущему и следующему кварталу
    df_base = df[df["quarter_idx"] == q_idx]
    df_next = df[df["quarter_idx"] == q_idx + 1]

    base = df_base["genotype"].value_counts()
    next_gen = df_next["genotype"].value_counts()

    # Векторное деление численности генотипов
    lambdas_series = next_gen.divide(base, fill_value=0)
    
    full_index = range(32)
    return lambdas_series.reindex(full_index, fill_value=0.0).tolist()


def delta_fitness_quarterly(q_idx: int, df):
    lambdas = survivability_quarterly(q_idx, df)
    max_surv = max(lambdas)
    if max_surv == 0:
        return 0.0

    Q = df[df["quarter_idx"] == q_idx + 1].shape[0]
    P = df[df["quarter_idx"] == q_idx].shape[0]
    if P == 0:
        return 0.0

    return Q / (P * max_surv)


def fitness_variance_quarterly(q_idx: int, df):
    df_quarter = df[df["quarter_idx"] == q_idx]
    P = len(df_quarter)
    if P == 0:
        return 0.0

    W_0 = delta_fitness_quarterly(q_idx, df)
    if W_0 == 0:
        return 0.0

    lambdas = survivability_quarterly(q_idx, df)
    max_surv = max(lambdas)
    if max_surv == 0:
        return 0.0

    w_series = pd.Series([l / max_surv for l in lambdas], index=range(32))
    p_series = df_quarter["genotype"].value_counts().reindex(range(32), fill_value=0)
    p_series = p_series / P

    variance = (p_series * ((w_series - W_0) ** 2)).sum()
    return variance / W_0

```

```python

deltas = []
variances = []

for year in range(min_q, max_q):
    deltas.append(delta_fitness_quarterly(year, df))
    variances.append(fitness_variance_quarterly(year, df))


# 2. Линейная корреляция Пирсона
r_pearson, p_pearson = pearsonr(deltas, variances)

# Проверка гипотезы тестом Спирмена
rho, p_val = spearmanr(deltas, variances)


print(f"Результаты теста для жанра Classical")
print(f"Ранговая корреляция Пирсона: {r_pearson:.4f}")
print(f"Статистическая значимость (p-value): {p_pearson:.4f}")
print(f"Ранговая корреляция Спирмена: {rho:.4f}")
print(f"Статистическая значимость (p-value): {p_val:.4f}")

# Визуализация

slope, intercept = np.polyfit(deltas, variances, 1) # Находим коэффициенты прямой y = kx + b
x_range = np.linspace(min(deltas), max(deltas), 100)
plt.plot(x_range, slope * x_range + intercept, color='blue', linestyle='-', linewidth=2)

plt.scatter(deltas, variances)
#plt.axline((0, 0), (1, 1), color="red", linestyle="--", label="Линия Фишера (1:1)")
plt.xlabel("Delta Fitness")
plt.ylabel("Fitness Variance")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

plt.show()
```

Наконец, объединим данные по всем жанрам (поколение = 1 год) и посмотрим, имеется ли корреляция в общем случае.

```python
global_data = []

all_genres = df['genre'].unique()

# Перебираем каждый жанр по отдельности
for genre_name in all_genres:
    genre_df = df[df['genre'] == genre_name]
    
    for year in range(2015, 2025):
        
        if len(genre_df[genre_df['year'] == year]) > 0 and len(genre_df[genre_df['year'] == year + 1]) > 0:
            
            d_fit = delta_fitness(year, genre_df)
            f_var = fitness_variance_fast(year, genre_df)
            
            # Исключаем артефакты (например, если из-за деления на 0 вернулся 0)
            if d_fit > 0 and f_var > 0:
                global_data.append({
                    'genre': genre_name,
                    'year': year,
                    'delta_fitness': d_fit,
                    'fitness_variance': f_var
                })

summary_df = pd.DataFrame(global_data)

# Ранговая корреляция Спирмена
rho_spearman, p_spearman = spearmanr(summary_df['delta_fitness'], summary_df['fitness_variance'])

# Линейная корреляция Пирсона
r_pearson, p_pearson = pearsonr(summary_df['delta_fitness'], summary_df['fitness_variance'])


print(f"Корреляция Пирсона (линейная):  r = {r_pearson:.4f}, p-value = {p_pearson:.4e}")
print(f"Корреляция Спирмена (ранговая): rho = {rho_spearman:.4f}, p-value = {p_spearman:.4e}\n")

# График с линией тренда
plt.figure(figsize=(10, 7))

# Рисуем жанры
all_genres = summary_df['genre'].unique()
for genre_name in all_genres:
    genre_subset = summary_df[summary_df['genre'] == genre_name]
    plt.scatter(genre_subset['delta_fitness'], genre_subset['fitness_variance'], label=genre_name, alpha=0.6)


x = summary_df['delta_fitness']
y = summary_df['fitness_variance']
slope, intercept = np.polyfit(x, y, 1) # Находим коэффициенты прямой y = kx + b
x_range = np.linspace(x.min(), x.max(), 100)
plt.plot(x_range, slope * x_range + intercept, color='blue', linestyle='-', linewidth=2, 
         label=f'Реальный тренд (y = {slope:.3f}x + {intercept:.3f})')

# Идеальная теоретическая прямая 1:1 по теореме Фишера
#plt.axline((0, 0), (1, 1), color='red', linestyle='--', linewidth=1.5, label='Теория Фишера (1:1)')

plt.title('Проверка теоремы Фишера: Пирсон & Спирмен')
plt.xlabel('Скорость повышения приспособленности ($\\Delta W$)')
plt.ylabel('Генетическая дисперсия / $W_0$')

# Добавляем метрики прямо на график
stats_text = f"Pearson r: {r_pearson:.4f} (p < 10^-7)\nSpearman $\\rho$: {rho_spearman:.4f} (p < 10^-7)"
plt.gca().text(0.05, 0.95, stats_text, transform=plt.gca().transAxes,
               fontsize=11, verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

#### Выводы

Изменения приспособленности генотипов внутри жанра не являются сильно скореллированными с генетической дисперсией. p-значения, полученные для отдельных жанров, не позволяют отвергнуть нулевую гипотезу и принять теорему Фишера, однако на агрегированных данных коэффициенты Спирмена и Пирсона показывают умеренный уровень корреляции с $p<0.05$, что может быть объяснено увеличением выборки. Поэтому нельзя с уверенностью говорить о том, что в музыке действует ествественный отбор. 

К недостаткам нашего подхода можно отнести определение генотипа музыка через бинаризацию непрерывных признаков, так как при этом теряется довольно много информации о треке. Исследование может быть продолжено с использованием уравнения Прайса вместо формулы Фишера, поскольку она позволяет естественным образом работать с непрерывными признаками.
